```python
# Q1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
# Add a new column that is the product of the first two columns.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])
df['Product'] = df['A'] * df['B']
print(df)
```

```python
# Q2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers.
# Compute the row-wise and column-wise sum.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)),
                  columns=['A', 'B', 'C'])
print("Row-wise sum:\n", df.sum(axis=1))
print("Column-wise sum:\n", df.sum(axis=0))
```

```python
# Q3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
# Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])
df.loc[1, 'A'] = np.nan
df.loc[2, 'B'] = np.nan
df.loc[3, 'C'] = np.nan

print("Before filling:\n", df)
df = df.fillna(df.mean())
print("After filling:\n", df)
```

```python
# Q4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers.
# Introduce some NaN values. Drop the rows with any NaN values.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D'])
df.loc[1, 'A'] = np.nan
df.loc[3, 'C'] = np.nan

print("Before dropping:\n", df)
df = df.dropna()
print("After dropping:\n", df)
```

```python
# Q5. Create a DataFrame with two columns: Category and Value.
# Fill Category with random categories A, B, C and Value with random integers.
# Group by Category and compute the sum and mean of Value.

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(10, 100, size=10)
})

print(df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

```python
# Q6. Create a DataFrame with Product, Category, and Sales columns.
# Fill the DataFrame with random data. Group by Category and compute total sales.

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Product': np.random.choice(['Pen', 'Book', 'Bag'], size=8),
    'Category': np.random.choice(['Stationery', 'Accessories'], size=8),
    'Sales': np.random.randint(100, 1000, size=8)
})

print(df)
print(df.groupby('Category')['Sales'].sum())
```

```python
# Q7. Create two Pandas DataFrames with a common column.
# Merge the DataFrames using the common column.

import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ali', 'John', 'Sara']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 78]
})

result = pd.merge(df1, df2, on='ID')
print(result)
```

```python
# Q8. Create a Pandas DataFrame with a MultiIndex (hierarchical index).
# Perform basic indexing and slicing operations.

import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('A', 1), ('A', 2), ('B', 1), ('B', 2)],
    names=['Category', 'Number']
)

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("Complete DataFrame:\n", df)
print("Category A:\n", df.loc['A'])
print("Specific row:\n", df.loc[('B', 1)])
print("Slicing:\n", df.loc['A':'B'])
```

```python
# Q9. Create a DataFrame with MultiIndex consisting of Category and SubCategory.
# Fill it with random data and compute the sum of values for each combination.

import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'X', 'Y', 'Y'],
    'Value': np.random.randint(10, 100, size=6)
})

df = df.set_index(['Category', 'SubCategory'])
print(df.groupby(level=['Category', 'SubCategory']).sum())
```

```python
# Q10. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
# Apply a function that doubles the values of the DataFrame.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:\n", df)
print("Doubled DataFrame:\n", df.map(lambda x: x * 2))
```

```python
# Q11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers.
# Apply a lambda function to create a new column that is the sum of existing columns.

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)
print(df)
```
